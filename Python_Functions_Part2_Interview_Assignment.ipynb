````python
# write your code here
def total(*args):
  sum=0
  for i in args:
    sum += i
  return f'total sum is {sum}'

total(2,1)
total()# It will return 0 because we have initialize the sum as o and not given any arguments
````

## Section 2 — `**kwargs`: Variable-Length Keyword Arguments
`**kwargs` lets a function accept any number of **keyword** arguments, collected internally
as a dictionary.

**Q4.** Write a function `print_profile(**kwargs)` that prints each key-value pair it
receives, one per line, in the format `key --> value`. Call it as
`print_profile(name="Rohan", age=22, city="Pune")`.

````python
# write your code here
def print_profile(**kwargs):
  for i in kwargs.items():
    print(f'{i[0]} --> {i[1]}')

print_profile(name="Rohan", age=22, city="Pune")
````

**Q6.** Write a function `build_url(base, **params)` that builds a query string from
keyword arguments, e.g. `build_url("https://api.com/search", q="python", limit=10)` should
return something like `"https://api.com/search?q=python&limit=10"`.

````python
# write your code here
def build_url(base, **params):
  api = base + '?'
  for i in params.items():
    api += i[0]+'='+i[1]+'&'
  return api[0:-1]

build_url("https://api.com/search", q="python", limit="10")
````

- `*amounts` collects however many line-item prices are passed in
- `tax_rate` is a keyword-only-style default (passed as a normal default arg)
- `**discounts` can optionally contain named discounts like `loyalty=50, coupon=20` (flat
  amounts to subtract)

The function should: sum the amounts, subtract the total of all discounts, then add tax on
the result, and return the final rounded total. Call it with 3 item prices, a custom
`tax_rate`, and two discounts.

````python
# write your code here
def invoice_total(*amounts, tax_rate=0.18, **discounts):
  sum_of_amounts= sum(amounts)
  sum_of_discounts = sum(discounts.values())
  total_after_discounts=sum_of_amounts-sum_of_discounts
  total=total_after_discounts*(1+tax_rate)
  return round(total,2)

invoice_total(1650,100,500,40000,85000,tax_rate=0.18,loyalty=80,coupon=60)
````
